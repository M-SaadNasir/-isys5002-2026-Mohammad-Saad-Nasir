```python
# Weather Menu

print("Weather Menu:")
print("1. Check Temperature")
print("2. Check Humidity")
print("3. Check Wind Speed")
print("4. Exit")

choice = input("Enter your choice (1-4): ")

if choice == "1":
    temperature = int(input("Enter the temperature in °C: "))

    if temperature < 5:
        print("It's freezing - bundle up!")
    elif temperature < 15:
        print("It's cold - wear a jacket!")
    elif temperature < 25:
        print("It's warm - enjoy the day!")
    else:
        print("It's hot - stay cool!")

elif choice == "2":
    humidity = int(input("Enter the humidity (%): "))

    if humidity > 70:
        print("It's quite humid today.")
    elif humidity >= 40:
        print("The humidity is comfortable.")
    else:
        print("The air is dry.")

elif choice == "3":
    wind_speed = int(input("Enter the wind speed (km/h): "))

    if wind_speed > 40:
        print("It's very windy - be careful!")
    elif wind_speed >= 20:
        print("There is a moderate wind.")
    else:
        print("The wind is light.")

elif choice == "4":
    print("Exiting the program.")

else:
    print("Invalid choice. Please select a valid option.")
```
